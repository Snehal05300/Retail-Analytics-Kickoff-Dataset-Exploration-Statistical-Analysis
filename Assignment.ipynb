```python
import pandas as pd

# First DataFrame
df1 = pd.DataFrame({
    'id': [0, 1, 2, 3, 4],
    'location': ['USA', 'USA', 'USA', 'USA', 'USA'],
    'age': [24, 31, 29, 33, 36],
    'gender': ['M', 'F', 'F', 'M', 'F'],
    'is_senior': [False, True, False, False, True]
})

# Second DataFrame
df2 = pd.DataFrame({
    'employee_id': [0, 1, 2, 3, 4],
    'age': [24, 31, 29, 33, 36],
    'gender': ['M', 'F', 'F', 'M', 'F'],
    'popularity': [6, 4, 0, 7, 6]
})

# 1. Merge the two DataFrames
merged_df = pd.merge(
    df1,
    df2,
    left_on='id',
    right_on='employee_id'
)

print("Merged DataFrame:")
print(merged_df)

# 2. Find average popularity as per location
average_popularity = merged_df.groupby('location')['popularity'].mean()

print("\nAverage Popularity as per Location:")
print(average_popularity)
```

```python
import pandas as pd

df = pd.read_csv(r"C:\ITvedant\Data Manipulation\Netflix 1.csv")

print(df.columns)
```

```python
df["Boxoffice"] = (
    df["Boxoffice"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["Boxoffice"] = pd.to_numeric(df["Boxoffice"], errors="coerce")

box_office_hits = df[df["Boxoffice"] >= 2000000]

print("Number of Box-Office Hit Films:", len(box_office_hits))
```

```python
critically_acclaimed = df[df["IMDb Score"] >= 8]

print("Number of critically acclaimed movies:", len(critically_acclaimed))
```

```python
critically_acclaimed_hits = df[
    (df["IMDb Score"] >= 8) &
    (df["Boxoffice"] >= 2000000)
]

print(critically_acclaimed_hits[["Title", "IMDb Score", "Boxoffice"]])
```

```python
acclaimed_awards = df[
    (df["IMDb Score"] >= 8) &
    (df["Awards Nominated For"] >= 5)
]

acclaimed_awards = acclaimed_awards.sort_values(
    "Awards Nominated For",
    ascending=False
)

print(acclaimed_awards[[
    "Title",
    "IMDb Score",
    "Awards Nominated For"
]])
```

```python
us_acclaimed_series = df[
    (df["IMDb Score"] >= 8) &
    (df["Series or Movie"] == "Series") &
    (df["Country Availability"].str.contains(
        "United States",
        case=False,
        na=False
    ))
]

print(us_acclaimed_series[[
    "Title",
    "IMDb Score",
    "Country Availability"
]])
```

```python
films_no_awards = df[
    (df["IMDb Votes"] >= 100) &
    (df["Awards Nominated For"] == 0)
]

print(films_no_awards[[
    "Title",
    "IMDb Votes",
    "Awards Nominated For"
]])
```

```python
df["Release Date"] = pd.to_datetime(
    df["Release Date"],
    errors="coerce"
)

df["Netflix Release Date"] = pd.to_datetime(
    df["Netflix Release Date"],
    errors="coerce"
)

df["Netflix_Release_Difference"] = (
    df["Netflix Release Date"] - df["Release Date"]
).dt.days
```

```python
popular_films = df[
    (df["Netflix_Release_Difference"] >= 0) &
    (df["Netflix_Release_Difference"] <= 730)
]

print(popular_films[[
    "Title",
    "Release Date",
    "Netflix Release Date",
    "Netflix_Release_Difference"
]])
```

```python
popular_canada = popular_films[
    popular_films["Country Availability"].str.contains(
        "Canada",
        case=False,
        na=False
    )
]

print(popular_canada[[
    "Title",
    "Country Availability"
]])
```

```python
df["Runtime_Minutes"] = (
    df["Runtime"]
    .str.extract(r"(\d+)")
    [0]
    .astype(float)
)
```

```python
movies_one_hour = df[
    (df["Series or Movie"] == "Movie") &
    (df["Runtime_Minutes"] >= 60)
]

print(movies_one_hour[[
    "Title",
    "Runtime"
]])
```

```python
print(df["Series or Movie"].unique())
```

```python
df["Language_Count"] = (
    df["Languages"]
    .fillna("")
    .str.split(",")
    .str.len()
)
```

```python
multi_language_series = df[
    (df["Series or Movie"] == "Series") &
    (df["Language_Count"] > 1)
]

print(multi_language_series[[
    "Title",
    "Languages"
]])
```

```python
df["Genre_Count"] = (
    df["Genre"]
    .fillna("")
    .str.split(",")
    .str.len()
)
```

```python
japan_multi_genre = df[
    (df["Series or Movie"] == "Movie") &
    (df["Country Availability"].str.contains(
        "Japan",
        case=False,
        na=False
    )) &
    (df["Genre_Count"] > 1)
]

print(japan_multi_genre[[
    "Title",
    "Genre",
    "Country Availability"
]])
```

```python
df["Total_Rating"] = (
    df["Hidden Gem Score"] +
    df["IMDb Score"]
)
```

```python
high_total_rating = df[
    df["Total_Rating"] >= 15
]

print(high_total_rating[[
    "Title",
    "Hidden Gem Score",
    "IMDb Score",
    "Total_Rating"
]])
```

```python
japan_drama_series = df[
    (df["Series or Movie"] == "Series") &
    (df["Genre"].str.contains(
        "Drama",
        case=False,
        na=False
    )) &
    (df["Country Availability"].str.contains(
        "Japan",
        case=False,
        na=False
    ))
]

print(japan_drama_series[[
    "Title",
    "Genre",
    "Country Availability"
]])
```

```python
directed_by_c = df[
    df["Director"].fillna("").str.startswith("C", na=False)
]

print(directed_by_c[[
    "Title",
    "Director"
]])
```

```python
christopher_nolan = df[
    df["Director"].fillna("").str.contains(
        "Christopher Nolan",
        case=False,
        na=False
    )
]

print(christopher_nolan[[
    "Title",
    "Director"
]])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\ITvedant\Data Manipulation\historical_automobile_sales.csv")
```

```python
yearly_sales = df.groupby("Year")["Automobile_Sales"].sum()

yearly_sales.plot(kind="line", marker="o", figsize=(10, 5))
plt.title("Automobile Sales Fluctuation Over the Years")
plt.xlabel("Year")
plt.ylabel("Automobile Sales")
plt.grid(True)
plt.show()
```

```python
vehicle_sales = df.groupby(
    ["Year", "Vehicle_Type"]
)["Automobile_Sales"].sum().unstack()

vehicle_sales.plot(kind="line", marker="o", figsize=(12,6))
plt.title("Automobile Sales")
plt.xlabel("Year")
plt.ylabel("Automobile Sales")
plt.legend(title="Vehicle Type")
plt.grid(True)
plt.show()
```

```python
recession_sales = df.groupby(
    ["Recession", "Vehicle_Type"]
)["Automobile_Sales"].sum().unstack()

recession_sales.plot(kind="bar", figsize=(12,6))

plt.title("Vehicle Sales: Recession vs Non-Recession")
plt.xlabel("Recession Status")
plt.ylabel("Automobile Sales")
plt.xticks([0,1], ["Non-Recession", "Recession"], rotation=0)
plt.legend(title="Vehicle Type")
plt.grid(axis="y")
plt.show()
```

```python
recession_data =df[df["Recession"]==1]

price_sales = recession_data.groupby(
    "Price"
)["Automobile_Sales"].sum().reset_index()

plt.figure(figsize=(10,6))
plt.scatter(
    price_sales["Price"],
    price_sales["Automobile_Sales"]
)

plt.title("Average Vehicle price vs Automobile Sales During Recession")
plt.xlabel("Average Vehicle Price")
plt.ylabel("Automobile Sales")
plt.grid(True)
plt.show()
```

```python
advertising = df.groupby("Recession")["Advertising_Expenditure"].sum()

plt.figure(figsize=(7,7))

plt.pie(
    advertising,
    labels=["Non-Recession", "Recession"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Advertising Expenditure: Recession vs Non-Recession")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("C:\ITvedant\Data Manipulation\historical_automobile_sales (1).csv")

df.head()
```

```python
sns.kdeplot(
    data=df,
    x="Price",
    hue="City",
    fill=True
)

plt.title("Distribution of Price for Different Cities")
plt.xlabel("Price")
plt.ylabel("Density")
plt.show()
```

```python
sns.relplot(
    data=df,
    x="Year",
    y="GDP",
    hue="Recession",
    kind="line"
)

plt.title("GDP Over Time with Recession")
plt.xlabel("Year")
plt.ylabel("GDP")
plt.show()
```

```python
sns.jointplot(
    data=df,
    x="Price",
    y="Automobile_Sales",
    kind="scatter"
)

plt.show()
```

```python
corr = df[
    ["GDP", "Growth_Rate", "unemployment_rate"]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.swarmplot(
    data=df,
    x="Month",
    y="Automobile_Sales",
    hue="Vehicle_Type"
)


plt.title("Automobile Sales by Month")
plt.xlabel("Month")
plt.ylabel("Automobile Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x="Year",
    y="Consumer_Confidence"
)

plt.title("Consumer Confidence by Year")
plt.xlabel("Year")
plt.ylabel("Consumer Confidence")
plt.xticks(rotation=45)
plt.show()
```
