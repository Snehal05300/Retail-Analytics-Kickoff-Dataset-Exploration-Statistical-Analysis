```python
numeric_cols = [
    'Quantity_Available',
    'Units_Sold',
    'Revenue',
    'Shipping_Cost'
    ]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
num_cols = ['Quantity_Available', 'Units_Sold', 
			'Revenue','Customer_Satisfaction', 
            'Delivery_Time', 'Shipping_Cost']

for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
cat_cols = ['City', 'Product_Name','Supplier']

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df['Category'] = df['Category'].str.strip().str.title()

df['Category'].unique()
```

```python
df["Profit_Amount"] = (
    df["Revenue"] *
    df["Profit_Margin"] / 100
)

df[["Revenue",
    "Profit_Margin",
    "Profit_Amount"]].sample(5)
```

```python
df["Revenue_Category"] = np.where(
    df["Revenue"] < 5000,
    "Low Revenue",
    np.where(
        df["Revenue"] < 10000,
        "Medium Revenue",
        "High Revenue"
    )
)

df[["Revenue", "Revenue_Category"]].head()
```

```python
monthly_revenue = df.groupby('Month')['Revenue'].sum() 
plt.figure(figsize=(12,6)) 
plt.plot( monthly_revenue.index, monthly_revenue.values, marker='o', linewidth=2 ) 
plt.title('Monthly Revenue Trend') 
plt.xlabel('Month') 
plt.ylabel('Total Revenue') 
plt.grid(True) 
plt.show()
```
